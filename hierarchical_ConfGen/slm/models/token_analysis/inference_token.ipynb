```python
import torch
import numpy as np
import pandas as pd
from scipy.spatial.distance import pdist, squareform
```

```python
ground_coarse_token = []
ground_fine_token = []
for i in range(1, 6):
    checkpoint_path = f"/data/esmdiff_target/bpti_kinetic_cluster/bpti_ptm/bpti_{i}.ptm"
    checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'))
    ground_coarse_token.append(checkpoint['small_codebook'])
    ground_fine_token.append(checkpoint['large_codebook'])
```

```python
target_indices = [2,5,6,7,8,9,10,11,13,14,15,16,17,18,19,25,35,36,38,39,40,41,42,43,52,54,55,56,57]
```

```python
checkpoint_path = f"/esmdiff_var/slm/models/residual_120K/bpti_beam_4_3_3/bpti_beam_token.pth"
checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'))
```

```python
matching_tokens_1 = []
for i in range(100):
    matches = (checkpoint[(0, 'result')][i][target_indices] == ground_coarse_token[0][target_indices])
    ratio = matches.float().mean()
    matching_tokens_1.append(ratio)
print(f"max : {max(matching_tokens_1)} \n min : {min(matching_tokens_1)}\n mean : {sum(matching_tokens_1) / len(matching_tokens_1)}\n")
```
