```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV 파일 경로
csv_path = '/hierarchical_ConfGen/logs/MaskedDiffusionLanguageModeling/runs/2025-12-15_10-47-00/csv/version_0/metrics.csv'  # 여기에 실제 파일 경로 입력

# CSV 읽기
df = pd.read_csv(csv_path)
```

```python

plt.figure(figsize=(8, 5))
plt.plot(df['step'], df['train/loss_step'], marker='o', label='Train Loss')
plt.xlabel('Step')
plt.ylabel('Train Loss')
plt.title('Train Loss per 50 step')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

N = 100  # 묶을 단위
grouped_loss = df['train/loss_step'].groupby(df.index // N).mean()
grouped_step = df['step'].groupby(df.index // N).mean()  # 평균 step 값 사용

# 그래프 그리기
plt.figure(figsize=(8, 5))
plt.plot(grouped_step, grouped_loss, marker='o', label='Train Loss (avg over 10 steps)')
plt.xlabel('Step')
plt.ylabel('Train Loss')
plt.title('5000-step Average')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
#plt.plot(df['step'], df['val/loss'].interpolate(), marker='o', linestyle='-')
plt.plot(df['step'], df['val/loss'], marker='o', linestyle='-', label='Val Loss')
plt.xlabel('Step')
plt.ylabel('val Loss')
plt.title('Val Loss')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
